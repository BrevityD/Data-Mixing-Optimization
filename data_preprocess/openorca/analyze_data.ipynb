```python
import json
import pandas as pd
from collections import defaultdict
from tqdm import tqdm
import os
import random
from pathlib import Path

base_path = Path("../.." ).resolve()
data_dir = base_path / "data"
```

## Delete irrelevant domains

```python
id_token_counts = {}
total_tokens = 0
output_path = data_dir / "orca_copy.json"
file_path = output_path

with output_path.open("r", encoding="utf-8") as read_file:
    ds = json.load(read_file)

for data in ds:
    if 'id' in data and 'input_len' in data and 'output_len' in data:
        # Calculate total tokens for the ID
        token_count = data['input_len'] + data['output_len']
        id_token_counts[data['id']] = id_token_counts.get(data['id'], 0) + token_count
        total_tokens += token_count


# Calculate proportions
id_proportions = {id_: count / total_tokens for id_, count in id_token_counts.items()}

# Print the proportions
print("Proportions of total tokens for each ID:")
for id_, proportion in id_proportions.items():
    print(f"ID: {id_}, Proportion: {proportion:.4f}")
```
