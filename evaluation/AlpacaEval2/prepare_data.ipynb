```python
import datasets
import json

eval_set = datasets.load_dataset("tatsu-lab/alpaca_eval", "alpaca_eval")["eval"]

print(eval_set[0])
```

```python
all_data_items = []
path = "/mbz/users/liyuan/datasets"
# Iterate over each item in the test split of the dataset
for item in eval_set:
    all_data_items.append(item)

# Define the path to the output file
output_file = f'{path}/eval/AlpacaEval.json'

# Write the modified data items to a JSON file
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(all_data_items, f, ensure_ascii=False, indent=2)
```
