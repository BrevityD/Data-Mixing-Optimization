```python
import json
import re

import re
import json

# Function to extract the answer from a given text using the "#### X" format
def extract_answer_from_text(answer_text):
    """
    Extracts the answer from the given text formatted with '####' followed by
    a number, 'True', 'False', or 'Uncertain'.

    Parameters:
        answer_text (str): The text containing the answer.

    Returns:
        int or bool or str or None: The extracted value as an integer, boolean,
                                     string ('Uncertain'), or None if no match is found.
    """
    # Define the regex pattern to match '####' followed by a number, True, False, or Uncertain
    pattern = r'####\s*(\d+|True|False|Uncertain)'
    
    # Perform a case-insensitive search
    match = re.search(pattern, answer_text, re.IGNORECASE)
    
    if match:
        value_lower = match.group(1).strip().lower()
        
        if value_lower == 'true':
            return "True"
        elif value_lower == 'false':
            return "False"
        elif value_lower == 'uncertain':
            return 'Uncertain'  # You can choose to return None or a specific enum if preferred
    

# Function to calculate the correctness ratio by comparing predict and label fields
def calculate_correctness_ratio(file_path):
    correct_count = 0
    total_items = 0

    # Open and read the JSONL file
    with open(file_path, 'r') as file:
        for line in file:
            item = json.loads(line.strip())
            
            # Extract answers from predict and label
            predicted_answer = extract_answer_from_text(item.get("predict", ""))
            label_answer = item.get("label", "")
            
            # Check if the answers match
            if predicted_answer == label_answer:
                correct_count += 1
            total_items += 1

    # Calculate the correctness ratio
    correctness_ratio = correct_count / total_items if total_items > 0 else 0
    return correctness_ratio
```

```python
# Load the JSON file and calculate the correctness ratio

# Example usage
file_path = '/mbz/users/liyuan/LLaMA-Factory/results/FOLIO_validation/FOLIO_like_data_sft_1.0e-6_seed42/generated_predictions.jsonl'
correctness_ratio = calculate_correctness_ratio(file_path)
print(f"Correctness Ratio: {correctness_ratio:.2%}")
```
